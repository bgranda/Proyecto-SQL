```python
import os
import pandas as pd
import psycopg2
from dotenv import load_dotenv

load_dotenv("../.env")
con = psycopg2.connect(os.getenv("DATABASE_URL"))
cur = con.cursor()
```

```python
def sql_query(query):
    cur.execute(query)
    ans = cur.fetchall()
    names = [description[0] for description in cur.description]
    return pd.DataFrame(ans, columns=names)
```

```python
with open("../sql/crear_tablas.sql", encoding="utf-8") as f:
    script = f.read()

cur.execute(script)
con.commit()
print("Tablas creadas")
```

```python
sql_query("""
    SELECT table_name
    FROM information_schema.tables
    WHERE table_schema = 'public'
    ORDER BY table_name
""")
```
